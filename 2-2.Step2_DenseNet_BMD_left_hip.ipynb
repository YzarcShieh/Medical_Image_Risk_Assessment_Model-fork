# Step 2-2. DenseNet121 — DXA/BMD Left Hip

## 1. Overview

This notebook demonstrates training a DenseNet121 model **from scratch** using demonstration dxa/bmd left hip images for type 2 diabetes (T2D) classification and extracting image activation vectors for downstream multimodal prediction.

Four separate DenseNet training jobs are provided: ABD-US, BMD left hip, BMD right hip, and BMD spine. This notebook runs only the **BMD_left_hip** job.

## 2. Image Dimensions and Preprocessing

**All image dimensions in this notebook are given as Height × Width (H × W).**

| Image stage | Height × Width | Input channels |
|---|---|---|
| Stored BMD_left_hip demonstration JPG | **299 × 250 pixels** | RGB file |
| DenseNet121 model input | **299 × 250 pixels** | 1 (first RGB channel) |

The demonstration JPG files have already been prepared. No additional resizing or padding is performed during DenseNet data loading: the code uses the stored **H × W = 299 × 250** pixels. The original BMD training implementation selects the first image channel and scales pixel intensity to [0, 1]. Data augmentation is applied only during training.

**Original image preprocessing is not reproduced here.** Due to data privacy restrictions, original individual-level medical images cannot be publicly provided, so the original image quality control, ROI cropping, and padding procedures cannot be demonstrated in this notebook. They are described separately in the documentation.


## 3. Dataset Partition

The demonstration dataset contains 100 participants, divided into four mutually exclusive participant-level subsets:

| Dataset | Participants | Purpose |
|---|---:|---|
| Training1 | 32 | DenseNet121 training (updates model weights) |
| Training2 | 32 | Activation vector extraction for downstream models |
| Validation | 16 | Checkpoint selection by validation AUC |
| Testing | 20 | Activation vector extraction and downstream evaluation |

DenseNet121 is trained **only** on Training1, and the best checkpoint is selected using Validation. After selection, the same model extracts activation vectors for all four partitions without further weight updates. Each demonstration participant has one BMD_left_hip image.

## 4. Model Architecture and Hyperparameters

| Hyperparameter | Original implementation | GitHub demo |
|---|---|---|
| Architecture | MONAI DenseNet121 | MONAI DenseNet121 |
| Optimizer | Adam | Adam |
| Learning rate | 1 × 10⁻⁵ | **1 × 10⁻⁵** |
| Loss function | Cross-entropy | Cross-entropy |
| Batch size | 64 | **64** (adjustable) |
| Pretrained weights | Not used | **Not used** |
| Maximum training epochs | 500 | **20**, to shorten computation time |
| Validation | Every epoch | Every epoch |
| Model selection | Highest validation AUC | Highest validation AUC |

Training augmentation uses random rotation (±15°), flipping, and zooming (0.9–1.1). Validation and activation-vector extraction use no random augmentation. All controllable parameters appear at the beginning of the notebook.

## 5. GPU and Computational Requirements

GPU acceleration is recommended but **no specific GPU model or GPU count is required**. Use `GPU_INDEX` to select a visible CUDA GPU, and reduce `TRAIN_BATCH_SIZE` / `EVAL_BATCH_SIZE` if memory is limited. CPU execution is supported as a fallback. The separate `t2d-gpu` environment is described in the following Environment Setup section.

## 6. Outputs

This notebook saves a model checkpoint, epoch-by-epoch training history, run configuration, and a **1024-dimensional activation vector for each of the 100 BMD_left_hip images**. Generated outputs are written to `Outputs/Step2_DenseNet/BMD_left_hip/`; original files in `Demo_data/` are not modified.

The validation AUC is for **checkpoint selection** and should not be reported as the final multimodal T2D prediction performance. Published study performance cannot be inferred from the 100-person demonstration dataset.

````python
# ============================================================
# 1. User-defined Hyperparameters — BMD_left_hip DenseNet121
# ============================================================

from pathlib import Path
import json
import time
import platform

import numpy as np
import pandas as pd
from PIL import Image

import torch
import monai

from monai.networks.nets import DenseNet121
from monai.utils import set_determinism


# ============================================================
# 1.1 Project and Dataset Configuration
# ============================================================

PROJECT_ROOT = Path.cwd().resolve()

MODALITY = "BMD_left_hip"

IMAGE_DIR = (
    PROJECT_ROOT
    / "Demo_data/1-2.Sample_image/BMD_left_hip"
)

SAMPLE_INFO_FILE = (
    PROJECT_ROOT
    / "Demo_data/1-1.Sample_information/sample_information.csv"
)

OUTPUT_DIR = (
    PROJECT_ROOT
    / "Outputs/Step2_DenseNet/BMD_left_hip"
)


# ============================================================
# 1.2 Image Configuration
# ============================================================

# Image dimensions: Height x Width
# Stored demonstration images: 299 x 250 pixels
SOURCE_IMAGE_HW = (299, 250)

# DenseNet121 input: 299 x 250 pixels (no resizing)
INPUT_HW = (299, 250)

# Use the first RGB channel (single-channel model input)
IN_CHANNELS = 1

# Binary classification: Control / T2D
NUM_CLASSES = 2


# ============================================================
# 1.3 Model Hyperparameters
# ============================================================

MODEL_NAME = "DenseNet121"

# Original implementation: training from scratch
PRETRAINED = False

# Original study: 500 epochs
# GitHub demonstration: 20 epochs to reduce computation time
MAX_EPOCHS = 20

# Learning rate from the original DenseNet implementation
LEARNING_RATE = 1e-5

# Optimizer and loss
OPTIMIZER = "Adam"
LOSS_FUNCTION = "CrossEntropyLoss"

# Batch size
TRAIN_BATCH_SIZE = 64
EVAL_BATCH_SIZE = 64

# Validation every epoch
VAL_INTERVAL = 1

# Reproducibility
SEED = 0


# ============================================================
# 1.4 Data Augmentation
# ============================================================

# Rotation: +/- 15 degrees
ROTATION_DEGREES = 15
ROTATION_PROB = 0.5

# Random flip
FLIP_AXIS = 0
FLIP_PROB = 0.5

# Random zoom
ZOOM_RANGE = (0.9, 1.1)
ZOOM_PROB = 0.5


# ============================================================
# 1.5 GPU / Computational Configuration
# ============================================================

# Index of the GPU visible to the current process
GPU_INDEX = 0

# Allow CPU execution when CUDA is unavailable
ALLOW_CPU_FALLBACK = True

# Data loading
NUM_WORKERS = 2

# Optional mixed precision
# False follows the original implementation
USE_AMP = False


# ============================================================
# 1.6 Output Configuration
# ============================================================

CHECKPOINT_FILE = (
    OUTPUT_DIR / "BMD_left_hip_best_model.pth"
)

HISTORY_FILE = (
    OUTPUT_DIR / "BMD_left_hip_TrainingHistory.csv"
)

CONFIG_FILE = (
    OUTPUT_DIR / "BMD_left_hip_RunConfig.json"
)

FEATURE_FILE = (
    OUTPUT_DIR / "BMD_left_hip_ActivationVectors.csv"
)


# ============================================================
# 1.7 Device Selection
# ============================================================

if torch.cuda.is_available():

    if GPU_INDEX >= torch.cuda.device_count():
        raise ValueError(
            f"GPU_INDEX={GPU_INDEX}, but only "
            f"{torch.cuda.device_count()} GPUs are visible."
        )

    DEVICE = torch.device(f"cuda:{GPU_INDEX}")

else:

    if not ALLOW_CPU_FALLBACK:
        raise RuntimeError(
            "CUDA is unavailable and CPU fallback is disabled."
        )

    DEVICE = torch.device("cpu")


# Set random seed
set_determinism(seed=SEED)

# Verify input data
assert SAMPLE_INFO_FILE.is_file(), SAMPLE_INFO_FILE
assert IMAGE_DIR.is_dir(), IMAGE_DIR

# Create output directory
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 1.8 Display Configuration
# ============================================================

print("=" * 65)
print("BMD_left_hip DenseNet121 — DEMONSTRATION CONFIGURATION")
print("=" * 65)

print("Model:", MODEL_NAME)
print("Modality:", MODALITY)

print("\nImage dimensions (Height x Width)")
print("Stored images:", SOURCE_IMAGE_HW)
print("Model input:", INPUT_HW)
print("Input channels:", IN_CHANNELS)

print("\nTraining parameters")
print("Optimizer:", OPTIMIZER)
print("Learning rate:", LEARNING_RATE)
print("Epochs:", MAX_EPOCHS)
print("Batch size:", TRAIN_BATCH_SIZE)
print("Loss:", LOSS_FUNCTION)
print("Pretrained:", PRETRAINED)
print("Validation interval:", VAL_INTERVAL)
print("Seed:", SEED)

print("\nSoftware environment")
print("Python:", platform.python_version())
print("PyTorch:", torch.__version__)
print("MONAI:", monai.__version__)

print("\nComputational device")
print("Device:", DEVICE)

if DEVICE.type == "cuda":

    gpu_name = torch.cuda.get_device_name(DEVICE)

    gpu_memory = (
        torch.cuda.get_device_properties(DEVICE).total_memory
        / 1024**3
    )

    print("GPU:", gpu_name)
    print("GPU memory (GiB):", round(gpu_memory, 2))

print("\nOutput directory:")
print(OUTPUT_DIR.relative_to(PROJECT_ROOT))

print("\nConfiguration completed.")
````

````python
# ============================================================
# 2. Load BMD_left_hip Images and Build DataLoaders
# ============================================================

from torch.utils.data import Dataset, DataLoader

from monai.transforms import (
    Compose,
    ScaleIntensity,
    RandRotate,
    RandFlip,
    RandZoom,
    EnsureType,
)


# ============================================================
# 2.1 Read Sample Information
# ============================================================

sample_info = pd.read_csv(
    SAMPLE_INFO_FILE,
    dtype={"Demo_ID": str}
)

assert len(sample_info) == 100
assert sample_info["Demo_ID"].is_unique

# Expected participant-level partitions
expected_splits = {
    "training1": 32,
    "training2": 32,
    "validation": 16,
    "testing": 20,
}

actual_splits = (
    sample_info["Dataset"]
    .value_counts()
    .to_dict()
)

assert actual_splits == expected_splits, (
    f"Unexpected dataset partitions: {actual_splits}"
)


# ============================================================
# 2.2 Prepare T2D Labels
# ============================================================

label_text = (
    sample_info["T2D"]
    .astype(str)
    .str.strip()
    .str.lower()
)

if set(label_text.unique()) <= {"0", "1"}:

    sample_info["Label"] = label_text.astype(int)

elif set(label_text.unique()) <= {"case", "control"}:

    sample_info["Label"] = (
        label_text
        .map({
            "control": 0,
            "case": 1
        })
        .astype(int)
    )

else:

    raise ValueError(
        f"Unexpected T2D labels: {label_text.unique().tolist()}"
    )


# ============================================================
# 2.3 Build Image-level Manifest
# ============================================================

records = []

for row in sample_info.itertuples(index=False):

    image_files = sorted(
        IMAGE_DIR.glob(
            f"{row.Demo_ID}_BMD_left_hip_*.jpg"
        )
    )

    expected_n = int(row.N_BMD_left_hip)

    assert len(image_files) == expected_n, (
        f"Image count mismatch for {row.Demo_ID}: "
        f"expected {expected_n}, found {len(image_files)}"
    )

    for image_file in image_files:

        records.append({
            "Demo_ID": row.Demo_ID,
            "Dataset": row.Dataset,
            "Label": int(row.Label),
            "Image_path": image_file,
            "Image_file": str(
                image_file.relative_to(PROJECT_ROOT)
            ),
        })


image_df = pd.DataFrame(records)

# Confirm that the stored BMD image sizes match the model input.
# Pillow returns (Width, Height); INPUT_HW is (Height, Width).
for file in image_df["Image_path"]:
    with Image.open(file) as im:
        actual_hw = (im.height, im.width)
    if actual_hw != SOURCE_IMAGE_HW:
        raise ValueError(
            f"Unexpected image dimensions for {file.name}: "
            f"H x W = {actual_hw}, expected {SOURCE_IMAGE_HW}"
        )

assert not image_df.empty
assert len(image_df) == int(sample_info["N_BMD_left_hip"].sum())
assert len(image_df) == 100
assert image_df["Demo_ID"].is_unique

print("=" * 65)
print("BMD_left_hip IMAGE DATASET")
print("=" * 65)

print("Participants:", len(sample_info))
print("Total BMD_left_hip images:", len(image_df))

print("\nDataset distribution:")

display(
    image_df.groupby("Dataset")
    .agg(
        Participants=("Demo_ID", "nunique"),
        Images=("Image_file", "size")
    )
    .reindex(list(expected_splits))
)


# ============================================================
# 2.4 Define Image Transforms
# ============================================================

def build_transforms(training=False):

    transforms = [
        ScaleIntensity(),
    ]

    if training:

        transforms.extend([

            RandRotate(
                range_x=np.deg2rad(ROTATION_DEGREES),
                prob=ROTATION_PROB,
                keep_size=True,
            ),

            RandFlip(
                spatial_axis=FLIP_AXIS,
                prob=FLIP_PROB,
            ),

            RandZoom(
                min_zoom=ZOOM_RANGE[0],
                max_zoom=ZOOM_RANGE[1],
                prob=ZOOM_PROB,
                keep_size=True,
            ),

        ])

    transforms.append(EnsureType())

    return Compose(transforms)


train_transform = build_transforms(training=True)
eval_transform = build_transforms(training=False)


# ============================================================
# 2.5 Define BMD_left_hip Dataset
# ============================================================

class BMDDataset(Dataset):

    def __init__(self, dataframe, transform):

        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):

        return len(self.df)

    def __getitem__(self, index):

        row = self.df.iloc[index]

        with Image.open(row["Image_path"]) as img:

            # Select the first RGB channel as in retained BMD training code
            image = np.asarray(
                img.convert("RGB"),
                dtype=np.float32
            )[:, :, 0].copy()

        # Add channel dimension
        # Height x Width -> Channel x Height x Width
        image = image[None, :, :]

        # Intensity scaling and augmentation; no resizing
        image = self.transform(image)

        return (
            image,
            int(row["Label"]),
            row["Demo_ID"],
            row["Image_file"],
        )


# ============================================================
# 2.6 Build DataLoaders
# ============================================================

def get_split(split):

    return image_df.loc[
        image_df["Dataset"] == split
    ].reset_index(drop=True)


def build_loader(dataset, batch_size, shuffle=False):

    return DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        num_workers=NUM_WORKERS,
        pin_memory=(DEVICE.type == "cuda"),
    )


train_ds = BMDDataset(
    get_split("training1"),
    train_transform
)

val_ds = BMDDataset(
    get_split("validation"),
    eval_transform
)

train_loader = build_loader(
    train_ds,
    TRAIN_BATCH_SIZE,
    shuffle=True
)

val_loader = build_loader(
    val_ds,
    EVAL_BATCH_SIZE,
    shuffle=False
)


# ============================================================
# 2.7 Verify Model Input
# ============================================================

image, label, demo_id, image_file = train_ds[0]

assert tuple(image.shape) == (
    IN_CHANNELS,
    INPUT_HW[0],
    INPUT_HW[1],
)

print("\n" + "=" * 65)
print("DATALOADER VERIFICATION")
print("=" * 65)

print("Demo_ID:", demo_id)
print("T2D label:", label)
print("Model input (C x H x W):", tuple(image.shape))

print(
    "Pixel intensity range:",
    round(float(image.min()), 4),
    "to",
    round(float(image.max()), 4)
)

print("\nDataLoaders ready.")
````

````python
# ============================================================
# 3. Train BMD_left_hip DenseNet121
# ============================================================

from sklearn.metrics import (
    roc_auc_score,
    accuracy_score
)


# ============================================================
# 3.1 Initialize Model
# ============================================================

model = DenseNet121(
    spatial_dims=2,
    in_channels=IN_CHANNELS,
    out_channels=NUM_CLASSES,
    pretrained=PRETRAINED,
).to(DEVICE)


# ============================================================
# 3.2 Optimizer and Loss Function
# ============================================================

criterion = torch.nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

amp_enabled = (
    USE_AMP and DEVICE.type == "cuda"
)

scaler = torch.cuda.amp.GradScaler(
    enabled=amp_enabled
)


# ============================================================
# 3.3 Save Run Configuration
# ============================================================

run_config = {

    "Modality": MODALITY,
    "Model": MODEL_NAME,

    "Stored_image_HW": list(SOURCE_IMAGE_HW),
    "Model_input_HW": list(INPUT_HW),

    "Training_participants": 32,
    "Validation_participants": 16,

    "Max_epochs_demo": MAX_EPOCHS,
    "Max_epochs_original": 500,

    "Optimizer": OPTIMIZER,
    "Learning_rate": LEARNING_RATE,
    "Loss": LOSS_FUNCTION,

    "Train_batch_size": TRAIN_BATCH_SIZE,
    "Eval_batch_size": EVAL_BATCH_SIZE,

    "Pretrained": PRETRAINED,
    "Validation_interval": VAL_INTERVAL,

    "Rotation_degrees": ROTATION_DEGREES,
    "Rotation_prob": ROTATION_PROB,
    "Flip_axis": FLIP_AXIS,
    "Flip_prob": FLIP_PROB,

    "Zoom_range": list(ZOOM_RANGE),
    "Zoom_prob": ZOOM_PROB,

    "Seed": SEED,
    "Num_workers": NUM_WORKERS,
    "Use_AMP": USE_AMP,

    "Device": str(DEVICE),

    "GPU_name": (
        torch.cuda.get_device_name(DEVICE)
        if DEVICE.type == "cuda"
        else None
    ),

    "PyTorch_version": torch.__version__,
    "MONAI_version": monai.__version__,
}

CONFIG_FILE.write_text(
    json.dumps(
        run_config,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)


# ============================================================
# 3.4 Training and Validation
# ============================================================

history = []

best_auc = float("-inf")
best_epoch = None

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats(DEVICE)


for epoch in range(1, MAX_EPOCHS + 1):

    epoch_start = time.perf_counter()

    # --------------------------------------------------------
    # Training1
    # --------------------------------------------------------

    model.train()

    train_loss_sum = 0.0
    train_n = 0

    for images, labels, _, _ in train_loader:

        images = images.to(
            DEVICE,
            dtype=torch.float32
        )

        labels = labels.to(
            DEVICE,
            dtype=torch.long
        )

        optimizer.zero_grad()

        with torch.cuda.amp.autocast(
            enabled=amp_enabled
        ):

            logits = model(images)

            loss = criterion(
                logits,
                labels
            )

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        batch_n = len(labels)

        train_loss_sum += (
            loss.item() * batch_n
        )

        train_n += batch_n

    train_loss = (
        train_loss_sum / train_n
    )


    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    val_auc = np.nan
    val_accuracy = np.nan

    if epoch % VAL_INTERVAL == 0:

        model.eval()

        val_labels = []
        val_probs = []

        with torch.no_grad():

            for images, labels, _, _ in val_loader:

                images = images.to(
                    DEVICE,
                    dtype=torch.float32
                )

                logits = model(images)

                probs = torch.softmax(
                    logits,
                    dim=1
                )[:, 1]

                val_probs.extend(
                    probs.cpu().numpy().tolist()
                )

                val_labels.extend(
                    labels.numpy().tolist()
                )

        if len(set(val_labels)) != 2:
            raise ValueError(
                "Validation data must contain both T2D classes."
            )

        val_auc = roc_auc_score(
            val_labels,
            val_probs
        )

        val_predictions = (
            np.asarray(val_probs) >= 0.5
        ).astype(int)

        val_accuracy = accuracy_score(
            val_labels,
            val_predictions
        )

        # ----------------------------------------------------
        # Save best checkpoint
        # ----------------------------------------------------

        if val_auc > best_auc:

            best_auc = val_auc
            best_epoch = epoch

            torch.save(
                model.state_dict(),
                CHECKPOINT_FILE
            )


    # --------------------------------------------------------
    # Training History
    # --------------------------------------------------------

    epoch_seconds = (
        time.perf_counter() - epoch_start
    )

    history.append({

        "Epoch": epoch,

        "Training_loss": train_loss,

        "Validation_AUC": val_auc,

        "Validation_accuracy": val_accuracy,

        "Epoch_seconds": epoch_seconds,

    })

    pd.DataFrame(history).to_csv(
        HISTORY_FILE,
        index=False
    )

    print(
        f"Epoch {epoch:02d}/{MAX_EPOCHS} | "
        f"Loss={train_loss:.4f} | "
        f"Val AUC={val_auc:.4f} | "
        f"Best AUC={best_auc:.4f} | "
        f"Time={epoch_seconds:.1f}s",
        flush=True
    )


# ============================================================
# 3.5 Training Summary
# ============================================================

print("\n" + "=" * 65)
print("BMD_left_hip DENSENET TRAINING COMPLETED")
print("=" * 65)

print("Optimizer:", OPTIMIZER)
print("Learning rate:", LEARNING_RATE)
print("Completed epochs:", len(history))

print("\nBest epoch:", best_epoch)
print("Best validation AUC:", round(best_auc, 4))

if DEVICE.type == "cuda":

    peak_memory = (
        torch.cuda.max_memory_allocated(DEVICE)
        / 1024**3
    )

    print(
        "Peak allocated GPU memory (GiB):",
        round(peak_memory, 3)
    )

print(
    "\nBest checkpoint:",
    CHECKPOINT_FILE.relative_to(PROJECT_ROOT)
)

print(
    "Training history:",
    HISTORY_FILE.relative_to(PROJECT_ROOT)
)

print("\nTraining completed successfully.")
````

````python
# ============================================================
# 4. Extract BMD_left_hip Activation Vectors
# ============================================================

# ============================================================
# 4.1 Load Best DenseNet Checkpoint
# ============================================================

if not CHECKPOINT_FILE.is_file():
    raise FileNotFoundError(CHECKPOINT_FILE)

model.load_state_dict(
    torch.load(
        CHECKPOINT_FILE,
        map_location=DEVICE
    )
)

model.eval()


# ============================================================
# 4.2 Extract Features from All Four Partitions
# ============================================================

feature_rows = []

splits = [
    "training1",
    "training2",
    "validation",
    "testing"
]

for split in splits:

    dataset = BMDDataset(
        get_split(split),
        eval_transform
    )

    loader = build_loader(
        dataset,
        EVAL_BATCH_SIZE,
        shuffle=False
    )

    with torch.no_grad():

        for images, labels, demo_ids, filenames in loader:

            images = images.to(
                DEVICE,
                dtype=torch.float32
            )

            # DenseNet feature maps
            feature_maps = model.features(images)

            # Extract activation vectors immediately
            # before the final linear classification layer.
            #
            # ReLU -> Global Average Pooling -> Flatten
            vectors = model.class_layers[:-1](
                feature_maps
            )

            # Final classification layer
            logits = model.class_layers[-1](
                vectors
            )

            probabilities = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            vectors = vectors.cpu().numpy()

            probabilities = (
                probabilities.cpu().numpy()
            )

            for i in range(len(demo_ids)):

                row = {

                    "Demo_ID": demo_ids[i],

                    "Dataset": split,

                    "Image_file": filenames[i],

                    "T2D": int(labels[i]),

                    "Predicted_Probability": float(
                        probabilities[i]
                    ),

                }

                row.update({

                    f"AV_{j:04d}": float(value)

                    for j, value in enumerate(
                        vectors[i]
                    )

                })

                feature_rows.append(row)


# ============================================================
# 4.3 Construct Activation Vector DataFrame
# ============================================================

activation_df = pd.DataFrame(
    feature_rows
)

activation_columns = [
    c for c in activation_df.columns
    if c.startswith("AV_")
]


# ============================================================
# 4.4 Validation
# ============================================================

assert len(activation_df) == len(image_df), (
    "Activation vector count does not match image count."
)

assert len(activation_columns) == 1024, (
    f"Expected 1024 activation dimensions, "
    f"found {len(activation_columns)}"
)

assert (
    activation_df[activation_columns]
    .notna()
    .all()
    .all()
), "Missing activation vector values detected."

assert activation_df["Image_file"].is_unique
assert activation_df["Demo_ID"].is_unique


# ============================================================
# 4.5 Save Activation Vectors
# ============================================================

activation_df.to_csv(
    FEATURE_FILE,
    index=False
)


# ============================================================
# 4.6 Output Summary
# ============================================================

print("=" * 65)
print("BMD_left_hip ACTIVATION VECTOR EXTRACTION")
print("=" * 65)

print("Total images:", len(activation_df))

print(
    "Unique participants:",
    activation_df["Demo_ID"].nunique()
)

print(
    "Activation dimensions:",
    len(activation_columns)
)

print("\nImage counts by dataset:")

print(
    activation_df["Dataset"]
    .value_counts()
    .reindex(splits)
    .to_string()
)

print(
    "\nOutput:",
    FEATURE_FILE.relative_to(PROJECT_ROOT)
)

display(
    activation_df[
        [
            "Demo_ID",
            "Dataset",
            "Image_file",
            "Predicted_Probability",
            "AV_0000",
            "AV_0001",
            "AV_0002",
        ]
    ].head()
)

print("\nActivation vector extraction completed.")
````
